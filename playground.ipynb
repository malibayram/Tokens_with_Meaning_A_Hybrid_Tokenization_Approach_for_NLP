```python
from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer

model = SentenceTransformer("alibayram/downstream-random-init")
model.save_pretrained("./model")
tabi_tokenizer = AutoTokenizer.from_pretrained(
        "alibayram/tabi-random-init", use_fast=False
    )
print(f"TabiBERT tokenizer vocab size: {tabi_tokenizer.vocab_size}")
tabi_tokenizer.save_pretrained("./model")
model = SentenceTransformer("./model")
model.push_to_hub("alibayram/tabi-random")
# remove ./model
import shutil
shutil.rmtree("./model")
```
